```python


import numpy as np
import networkx as nx
from scipy import linalg
from netneurotools import metrics

def pid_terms(X1, X2, Y):
    """Partial Information Decomposition using dit library"""
    import dit
    from dit.multivariate import coinformation
    
    # Create joint distribution
    outcomes = list(zip(X1, X2, Y))
    d = dit.Distribution(outcomes)
    
    # Compute PID components (requires specific PID measure)
    from dit.pid import PID_WB
    pid = PID_WB(d, [[0], [1]], [2])
    
    return {
        'unique_X1': pid[((0,), (1,))],
        'unique_X2': pid[((1,), (0,))],
        'synergy': pid[((0, 1),)],
        'redundancy': pid[((0,), (1,))]
    }
```
