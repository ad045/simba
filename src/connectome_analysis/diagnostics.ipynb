```python
import numpy as np 

connectomes = np.load("/Users/adrian/Documents/01_projects/14_4D_lab/data/preprocessed/01_first_analysises/connectomes_weighted_68x68.npy") # , allow_pickle=True).item()
connectomes.shape

connectome = connectomes[0]
connectome.shape
```

```python
# Check if connectome is connected
import numpy as np
from scipy.sparse import csr_matrix
from scipy.sparse.csgraph import connected_components

# Assuming 'connectome' is your weight matrix
n_components, labels = connected_components(csr_matrix(connectome), directed=False)
print(f"Number of connected components: {n_components}")  # Should be 1

# Check weight statistics
print(f"Weight range: [{connectome.min():.6f}, {connectome.max():.6f}]")
print(f"Non-zero weights: {(connectome != 0).sum()}")
print(f"Mean non-zero weight: {connectome[connectome != 0].mean():.6f}")
```
