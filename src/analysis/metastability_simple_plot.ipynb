```python
import numpy as np
import matplotlib.pyplot as plt
from pathlib import Path

import os
import sys

sys.path.insert(0, os.path.abspath('..'))
```

```python
# Loop through all the files in this folder for plotting
folder_name = Path("/Users/adrian/Documents/01_projects/14_4D_lab/14_4D_lab_code/output/metastability_debug/")

K_values = []
res_global = []
res_local_mean = []
for i, results_path in enumerate(folder_name.glob("*.npy")):
    data = np.load(results_path, allow_pickle=True).item()
    K_values.append(data["K_values"])
    res_global.append(data["global"])
    res_local_mean.append(data["local_mean"])
    
    if i >= 10:  # limit to first few files for clarity
        break
    
# K_values # should be the same for all files
K_values_for_plot = K_values[0]

res_global = np.array(res_global)
res_local_mean = np.array(res_local_mean)

plt.plot(K_values_for_plot, res_global.T, alpha=0.2) # , color='red')
plt.title("Global Metastability for {} random networks".format(res_global.shape[0]))
plt.xlabel("K values")
plt.ylabel("Global Metastability")
plt.show()
```
