```python
# generate 1 percent seeds for all animals: 

from netneurotools.networks import binarize_network

density = 1 # 0.5 # percent  (just vary this here and rerun the cell to create different density seed matrices)

all_seeds = []
# animal_id = 169 
for animal_id in range(connectomes.shape[0]):  
    conn = connectomes[animal_id]  # shape (100, 100)
    binarized_conn = binarize_network(conn, retain=density, keep_diag=False) # 2 percent
    all_seeds.append(binarized_conn)
all_seeds = np.array(all_seeds)  # shape (n_animals, n_regions, n_regions)

plt.imshow(all_seeds[0], cmap='hot', interpolation='nearest')
np.save(preprocessed_base_path / f"all_animal_seeds_density{density}.npy", all_seeds)
    # np.save(preprocessed_base_path / f"chimpanzee_{animal_id}_seed_density{"1p5"}.npy", binarized_consensus)
```
