```python
import numpy as np 
import matplotlib.pyplot as plt
import pandas as pd 
```

```python
df = pd.read_csv("/Users/adrian/Documents/01_projects/14_4D_lab/14_4D_lab_code/output/gnm/suarez_MaMI_dataset/60_generally_finer_search_animal_0/all_metrics_for_60_generally_finer_search_animal_0_updated.csv")
df_mc = pd.read_csv("/Users/adrian/Documents/01_projects/14_4D_lab/14_4D_lab_code/output/gnm/suarez_MaMI_dataset/60_generally_finer_search_animal_0/all_metrics_for_60_generally_finer_search_animal_0.csv")
```

```python
# merge both dfs on eta and gamma 
merged_df = pd.merge(df, df_mc, on=['eta', 'gamma'], suffixes=('_new', '_mc'))
merged_df.head()
```

```python
plt.scatter(merged_df['propagation_efficiency'], merged_df['mc_mean'], alpha=1, s=3, c=merged_df['MaxCriteria(DegreeKS, ClusteringKS, EdgeLengthKS, BetweennessKS)'], cmap='viridis')
plt.xlabel('Propagation Efficiency')
plt.colorbar()
plt.ylabel('Mean MC')
```

```python
merged_df[merged_df["propagation_efficiency"]>0.5]
```

```python
filtered_df = merged_df[merged_df["MaxCriteria(DegreeKS, ClusteringKS, EdgeLengthKS, BetweennessKS)"]>0.7] 
filtered_df = filtered_df[filtered_df["propagation_efficiency"]>0.5]
plt.scatter(filtered_df['propagation_efficiency'], 
            filtered_df['mc_mean'], 
            alpha=1, s=3, 
            c=filtered_df['MaxCriteria(DegreeKS, ClusteringKS, EdgeLengthKS, BetweennessKS)'], 
            cmap='viridis')
plt.xlabel('Propagation Efficiency')
plt.colorbar()
plt.ylabel('Mean MC')

# draw a trend line 
z = np.polyfit(filtered_df['propagation_efficiency'], filtered_df['mc_mean'], 1)
p = np.poly1d(z)
plt.plot(filtered_df['propagation_efficiency'], p(filtered_df['propagation_efficiency']), "r--")
```

```python
from netneurotools.modularity import consensus_modularity
import numpy as np 

A = np.load("/Users/adrian/Documents/01_projects/14_4D_lab/14_4D_lab_code/output/gnm/suarez_MaMI_dataset/60_generally_finer_search_animal_0/generated_networks/net_eta-0.673828125_gamma0.8398066163063049_ruleMatchingIndex.npy")[0]
A.shape
```

```python
from networkx.algorithms.community import modularity, louvain_communities, greedy_modularity_communities
import networkx as nx
```

```python
import time as time 

start_time = time.time()

G = nx.from_numpy_array(A)
louvain_comms = louvain_communities(G)
print(modularity(G, louvain_comms) if len(louvain_comms) > 1 else 0.0)

end_time = time.time()
print(f"Duration: {end_time - start_time} seconds")
```

```python
start_time = time.time()

comms = list(greedy_modularity_communities(G))
print(modularity(G, comms) if len(comms) > 1 else 0.0)

end_time = time.time()
print(f"Duration: {end_time - start_time} seconds") 
```
